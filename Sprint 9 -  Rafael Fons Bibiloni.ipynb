```python
# Nivel 1 - Ejercicio 4

# Diccionario inverso (con posibilidad de duplicados)
# Resulta que el cliente tiene una encuesta muy antigua que se almacena en un diccionario y los resultados los necesita a la inversa,
# es decir, intercambiando las claves y valores. 

encuesta = {
    "a":"pepe",
    "b":"no lo se",
    "c":7
}

def invertir_dicc(diccionario):
    claves = list(diccionario.keys())
    dicc_invertido ={}
    for i in range(len(diccionario)):
        dicc_invertido[diccionario[claves[i]]] = claves[i]

    return dicc_invertido
```

```python
# hacemos la funcion para contar palabras y hacer el diccionario

import string                   # me sirve para importar el abecedario en lugar de escribirlo

def contar_palabras_por_letra(texto):    
    # Limpiar el texto
    caracteres_limpios = []
    for ch in texto:
        if ch.isalpha():                        # esta funcion = True si el caracter es: a-z, ñ, á, etc.). Quito signo de puntuación
            caracteres_limpios.append(ch.lower())
        else:
            caracteres_limpios.append(" ")
    limpio = "".join(caracteres_limpios)        # genero un string uniendo los caracteres limpios y en minúsculas
    
    palabras = limpio.split()                   # separo las palabras en una lista
    
    # Diccionario principal: letra -> otro diccionario
    frecuencia = {}
    for letra in string.ascii_lowercase:        # importo el abecedario en minúscula
        frecuencia[letra] = {}                  #creo un diccionario vacío por cada letra del abecedario
        
    # Contar 
    for palabra in palabras:
        ch_inicial = palabra[0]                 # caracter inicial de cada palabra 
        # Si la palabra ya está, le sumamos 1
        if palabra in frecuencia[ch_inicial]:
            frecuencia[ch_inicial][palabra] += 1
        else:
            # Si es la primera vez que aparece, la inicializamos con 1
            frecuencia[ch_inicial][palabra] = 1
    
    # Ordenar el diccionario final 
    resultado = {}
    for ch_inicial in frecuencia.keys():            # no la ordeno porque se creo ya ordenada usando string
        if frecuencia[ch_inicial]:
            # Ordenamos las palabras dentro de cada ch_inicial
            resultado[ch_inicial] = dict(sorted(frecuencia[ch_inicial].items()))
    
    return resultado
```

```python
def conversor_to_float(lista):
    # recorremos los elementos y hacemos una lista
    lista_float = []
    lista_no_float = []

    for elem in lista:
        # si el elemento es una lista o tupla
        if isinstance(elem, (list, tuple)):
            for sub_elem in elem:
                try:
                    lista_float.append(float(sub_elem))
                except:
                    lista_no_float.append(sub_elem)

        # si el elemento es un valor normal
        else:
            try:
                lista_float.append(float(elem))
            except:
                lista_no_float.append(elem)

    return lista_float, lista_no_float
```

```python
def crear_contraseña(long, mayus, minus, num, signos):
    """
    # Generador de contraseñas:
    
    longitud (int): Longitud de la contraseña
    mayúsculas (bool = True): Si tiene que aparecer mayúsculas...string.ascii_uppercase
    minúsculas (bool = True): Si tiene que aparecer minúsculas...string.ascii_lowercase
    números (bool = True): Si tiene que aparecer números...string.digits
    signos (bool = False): Si tiene que aparecer caracteres especiales (,-$? o similares)...string.punctuation
    
    """
    
    from numpy import random
    import string
    
    # defino todos los posibles carateres que puedo elegir y los minimos carcateres que se requieren:
    posibles_ch = ""
    unicos_ch = ""              #defino el minimo de cada tipo para asegurar que cumple con los requisitos
    cant_ch_min = 0
    
    if mayus == True:
        cant_ch_min += 1
        posibles_ch += string.ascii_uppercase
        unicos_ch += random.choice(list(string.ascii_uppercase), 1)[0]                  #tengo que indicar el [0] porque numpy genera una lista
    if minus == True:
        cant_ch_min += 1
        posibles_ch += string.ascii_lowercase
        unicos_ch += random.choice(list(string.ascii_lowercase), 1)[0]
    if num == True:
        cant_ch_min += 1
        posibles_ch += string.digits
        unicos_ch += random.choice(list(string.digits), 1)[0]        
    if signos == True:
        cant_ch_min += 1
        posibles_ch += string.punctuation
        unicos_ch += random.choice(list(string.punctuation), 1)[0]
    
    
    # ajusto longitud de contraseña minima segun requisitos:
    if long < (cant_ch_min):
        print("ERROR: la longitud de la contraseña debe ser mayor a 4 caracteres")
    
    else:
        try:
            # unifico los posbles caracteres (minmos segun requisitos + resto de posibles carectares)   
            ch_finales = list( unicos_ch + ''.join(random.choice(list(posibles_ch), (long-cant_ch_min))) )
            
            # desordena la lista
            random.shuffle(ch_finales)             # shuffle no devuelve nada
            
            #uno los caracteres de la lista para obtener un string
            contraseña = ''.join(ch_finales)
            return contraseña 
        
        except:
            print("Los parametros ingresados no son correctos")
        
      
    
```

```python
# Importo el archivo "historic_partits.txt"

import pandas as pd

ruta = "/Users/rafafons/Library/Mobile Documents/com~apple~CloudDocs/2_Rafa/01-Capacitaciones/IT Academy/sprint 9/historic partits.txt"

df_historic_partits = pd.read_csv(ruta, sep="\t", header=None, names=["Local", "Resultado", "Visitante"])
df_historic_partits
```

```python
type(df_historic_partits.loc[3,"Resultado"])

# Convierto el campo resultado en 2 columnas de goles
for i in range(0,len(df_historic_partits)):       
    resultado = list(df_historic_partits.loc[i,"Resultado"])

    df_historic_partits.loc[i,"Goles local"] = int(resultado[0])
    df_historic_partits.loc[i,"Goles visitante"] = int(resultado[2])

#analizamos el resultado y colocamos puntajes
for i in range(0,len(df_historic_partits)):       
    # Caso 1: partidos ganados por local
    if df_historic_partits.loc[i,"Goles local"] > df_historic_partits.loc[i,"Goles visitante"]:
        df_historic_partits.loc[i,"Puntos local"] = 3
        df_historic_partits.loc[i,"Puntos visitante"] = 0
    #caso 2: partidos empatados
    elif df_historic_partits.loc[i,"Goles local"] == df_historic_partits.loc[i,"Goles visitante"]:
        df_historic_partits.loc[i,"Puntos local"] = 1
        df_historic_partits.loc[i,"Puntos visitante"] = 1
    #caso 3: partidas perdidos por local
    elif df_historic_partits.loc[i,"Goles local"] < df_historic_partits.loc[i,"Goles visitante"]:
        df_historic_partits.loc[i,"Puntos local"] = 0
        df_historic_partits.loc[i,"Puntos visitante"] = 3    

df_historic_partits
```

```python
# hago 2 tablas con los datos por equipo x fila (local y viistante) y luego las concateno 

locales = df_historic_partits[["Local", "Resultado", "Visitante", "Goles local", "Goles visitante", "Puntos local"]]
locales = locales.rename(columns={"Local":"Equipo", "Visitante":"Rival", "Goles local":"Goles convertidos",
                                  "Goles visitante": "Goles recibidos", "Puntos local":"Puntos"})

#calculo la cantidad de partidos ganados, empatados y perdidos
for i in range(0,len(locales)):       
    # Caso 1: partidos ganados por local
    if locales.loc[i,"Goles convertidos"] > locales.loc[i,"Goles recibidos"]:
        locales.loc[i,"Ganados"] = 1
    #caso 2: partidos empatados
    elif locales.loc[i,"Goles convertidos"] == locales.loc[i,"Goles recibidos"]:
        locales.loc[i,"Empatados"] = 1
    #caso 3: partidas perdidos por local
    elif locales.loc[i,"Goles convertidos"] < locales.loc[i,"Goles recibidos"]:
        locales.loc[i,"Perdidos"] = 1    

locales
```

```python
# hago la tabla de visitantes

visitantes = df_historic_partits[["Visitante", "Resultado", "Local", "Goles visitante", "Goles local", "Puntos visitante"]]
visitantes = visitantes.rename(columns={"Visitante":"Equipo", "Local":"Rival", "Goles local":"Goles recibidos",
                                  "Goles visitante": "Goles convertidos", "Puntos visitante":"Puntos"})
# invierto el resultado para que tenga sentido:
for i in range(0,len(visitantes)):       
    resultado = visitantes.loc[i,"Resultado"]
    visitantes.loc[i,"Resultado"] = resultado[::-1]

#calculo la cantidad de partidos ganados, empatados y perdidos
for i in range(0,len(locales)):       
    # Caso 1: partidos ganados por local
    if visitantes.loc[i,"Goles convertidos"] > visitantes.loc[i,"Goles recibidos"]:
        visitantes.loc[i,"Ganados"] = 1
    #caso 2: partidos empatados
    elif visitantes.loc[i,"Goles convertidos"] == visitantes.loc[i,"Goles recibidos"]:
        visitantes.loc[i,"Empatados"] = 1
    #caso 3: partidas perdidos por local
    elif visitantes.loc[i,"Goles convertidos"] < visitantes.loc[i,"Goles recibidos"]:
        visitantes.loc[i,"Perdidos"] = 1    
visitantes
```

```python
# tabla con todos los datos
locales_visitantes = pd.concat([locales, visitantes], ignore_index= True)
locales_visitantes
```

```python
# armo las consultas:

resumen = locales_visitantes.groupby(["Equipo"]).agg({"Ganados":"count", "Empatados":"count","Perdidos":"count",
                                                      "Goles convertidos": "sum", "Goles recibidos": "sum",
                                                       "Puntos": "sum"}).sort_values(["Puntos"], ascending=False).reset_index()
resumen
```
