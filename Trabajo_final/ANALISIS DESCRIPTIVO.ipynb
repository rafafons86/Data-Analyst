## 1.2 Tendencia de PIB

```python
# DATOS AGRUPADOS por año SIN 2020
data_1 = gastos_2023_pand.groupby("Año").agg({"PIB_MM_2023":"sum"}).round(0).reset_index()
data_2 = gastos_2023_pand.groupby("Año").agg({"% PIB 2023":"sum"}).round(4).reset_index()
data_3 = gastos_2023_pand.groupby("Año").agg({"% PIB 2023":"sum"}).round(4).reset_index()

# PERIODOS PIB actualizado a 2023
p1_pib = data_1[(data_1["Año"] <= 2007)]
p2_pib = data_1[(data_1["Año"] >= 2007)]

# Y: PIB actualizado a 2023
x1, y1 = p1_pib["Año"], p1_pib["PIB_MM_2023"]
x2, y2 = p2_pib["Año"], p2_pib["PIB_MM_2023"]

# REGRESIONES PIB
m1, b1 = np.polyfit(x1, y1, 1)
m2, b2 = np.polyfit(x2, y2, 1)

# LÍNEAS
p1_line = m1 * x1 + b1
p2_line = m2 * x2 + b2

# FORMULAS PARA TEXTO
formula1 = f"1995-2007: y = {m1:.0f}x + {b1:.0f}"
formula2 = f"2007-2023: y = {m2:.0f}x + {b2:.0f}"

# AGREGAR LÍNEAS AL GRÁFICO
fig = go.Figure()
# PIB anual actualizado a 2023
fig.add_trace(go.Scatter(
    x=data_1["Año"],
    y=data_1["PIB_MM_2023"],
    mode="lines+markers",
    name="PIB actualizado a 2023"
))
# % gastos sobre PIB (SIN AÑO 2020)
fig.add_trace(go.Scatter(
    x=data_2["Año"],
    y=data_2["% PIB 2023"],
    mode="lines+markers",
    name="% gasto s/PIB",
    yaxis="y2"
))
# tendencia PIB 1995-2007
fig.add_trace(go.Scatter(
    x=x1,
    y=p1_line,
    mode="lines",
    name="Tendencia PIB 1995-2007",
    line=dict(dash="dash", width=3)
))
# tendencia PIB 2007-2023
fig.add_trace(go.Scatter(
    x=x2,
    y=p2_line,
    mode="lines",
    name="Tendencia PIB 2007-2023",
    line=dict(dash="dash", width=3)
))
# Configuración del layout para eje secundario
fig.update_layout(
    title="Evolución del PIB y gasto del estado español (sin año 2020)",
    height=550,
    xaxis=dict(
        title="Año",
        dtick=1,                        # muestra cada dato unico, fuerza a
        tickfont=dict(size=12),
        range = [1994.5, 2023.5]),
    yaxis=dict(title="PIB en billones de €"),
    yaxis2=dict(
        title="% gasto estado sobre PIB",
        overlaying="y",
        side="right",
        range = [0.35, .6]),
    legend=dict(
        x=.8,
        y=1.2,
        xanchor="center",    #que punto de referencia se toma para el objeto visual
        yanchor="middle")
)
### NOTAS ###
# tendencia PIB 1995-2007
fig.add_annotation(
    x=2000,
    y=y1.iloc[-1]*.8,
    text=f"Var prom PIB: +{((y1.iloc[-1]/y1.iloc[0]-1) * 100):.0f}%",
    showarrow=False,
    font=dict(size=12, color="black")
)
# tendencia PIB 2007-2023
fig.add_annotation(
    x=2015,
    y=y2.iloc[0]*.95,   # valor en la escala de la recta (y2)
    text=f"Var prom PIB: +{((y2.iloc[-1] / y2.iloc[0] - 1) * 100):.0f}%",
    showarrow=False,
    font=dict(size=12, color="black")
)
fig.show()
```

```python
from plotly.subplots import make_subplots

# Definir los filtros y calcular las alturas
filtros_minarquistas = sorted(clasif_funciones["Clasif_minarquista"].unique())

# CalculO el número de filas (categorías) para cada filtro
num_filas = []
for clasificacion in filtros_minarquistas:
    num_cat = data_1[data_1["Clasif_minarquista"] == clasificacion]["Cod+Función"].nunique()
    num_filas.append(num_cat)

# Usamos una altura base (ej. 30 pixeles por fila)
base_height = 30
total_altura_requerida = sum(num_filas) * base_height

# Calcular alturas proporcionales para row_heights
row_heights = [(n * base_height) / total_altura_requerida for n in num_filas]

# Definir la altura total de la figura
fig_total_height = max(1000, total_altura_requerida + 150) # Altura mínima de 1000px + espacio para títulos/ejes

# Crear una figura con 4 filas y 1 columna, usando las alturas calculadas
fig = make_subplots(
    rows=len(filtros_minarquistas),
    cols=1,
    shared_xaxes=True,
    vertical_spacing=0.03, # Reducir el espacio vertical
    subplot_titles=filtros_minarquistas,
    row_heights=row_heights # 👈 CLAVE: Alturas proporcionales
)

# 2. Iterar y generar cada Heatmap
for i, clasificacion in enumerate(filtros_minarquistas):
    # ... Lógica de filtrado y normalización 
    data_filtrada = data_1[data_1["Clasif_minarquista"] == clasificacion].copy()
    data_pivot = data_filtrada.pivot(
        index="Cod+Función", 
        columns="Año", 
        values="Millones Euros 2023"
    ).sort_index(ascending=False)

    # NORMALIZO DATOS
    data_normalizada = data_pivot.div(data_pivot.sum(axis=1), axis=0)#.fillna(0)
    
    # 2.4. Crear el trazo (trace) de Heatmap
    heatmap_trace = go.Heatmap(
        z=data_normalizada.values,
        x=data_normalizada.columns.tolist(),
        y=data_normalizada.index.tolist(),
        colorscale="turbo",
        # -----------------------------------
        # Añadir esto para asegurar el vínculo con el coloraxis global
        coloraxis='coloraxis', 
        showscale=False # Desactivar la barra de color individual
    )
    
    # 2.5. Añadir el trazo al subplot
    fig.add_trace(heatmap_trace, row=i + 1, col=1)

# ⚠️ DEFINICIÓN DE LA ESCALA PERSONALIZADA ANTES DEL LAYOUT ⚠️
# ----------------------------------------------------
escala_personalizada = [
    [0, 'rgb(255, 255, 255)'],    #  Blanco
    #[0.25, 'rgb(30, 144, 255)'],   # Azul oscuro 
    [1, 'rgb(255, 99, 71)']      # Rojo claro
]

# 3. Configuración del Layout Global (Asegurando la Barra de Color)

fig.update_layout(
    title_text="Evolución del Gasto COFOG y por Clasificación Minarquista (Normalizado global)",
    height=fig_total_height, # Usar la altura total calculada
    # CONFIGURACIÓN CLAVE PARA LA BARRA DE COLOR GLOBAL
    coloraxis=dict(
        colorscale=escala_personalizada,#"turbo",
        colorbar=dict(
            title='% gasto',
            #titleside='right',
            # Ajuste de posición opcional para evitar colisiones
            x=1.05,
            yanchor="middle",
            len=0.9 # Hace la barra más larga
        )
    )
)

# 4. Ajustar los ejes X
for i in range(1, 5):
    fig.update_xaxes(dtick=1, row=i, col=1)

fig.show()
```

## Tendencia mediano plazo de gastos agrupados por minarquismo per PEA (1995-2007-2023)

```python
## DATOS ####
# hare el calculo per capita
data_1 = gastos_2023_pand.groupby(["Año","Clasif_minarquista"]).agg({"Millones Euros 2023":"sum"}).reset_index()
data_1 = data_1.merge(datos_macro["cant PEA"], on="Año", how="left")
data_1["Gasto_per_pea"] = round(data_1["Millones Euros 2023"]/data_1["cant PEA"]*1000000,0)

#pib per PEA

data_2 = pib_anual.merge(datos_macro["cant PEA"], on="Año", how="left")
data_2["PIB_per_pea"] = round(data_2["pib (MM € 2023)"]/data_2["cant PEA"]*1000000,0)
filtro_pandemia = data_2["Año"]!=2020
data_2 = data_2[filtro_pandemia]

# Definición de los períodos a analizar
PERIODOS = {
    "P1 (1995-2007)": (1995, 2007),
    "P2 (2007-2023)": (2007, 2023)
}

# Gráfico base (se mantiene igual)
fig = px.line(
    data_1,
    x="Año",
    y="Gasto_per_pea",
    color="Clasif_minarquista",
    markers=True
)
# agreggo PIB per PEA
fig.add_trace(go.Scatter(
    x=data_2["Año"],
    y=data_2["PIB_per_pea"],
    name="PIB per PEA (derecha)",
    mode="lines+markers",
    marker=dict(symbol="diamond", size=7),
    line=dict(dash="dash", width=3),
    yaxis="y2"
))

# Regresiones + anotaciones (MODIFICADO para doble período)
for grupo, df_group in data_1.groupby("Clasif_minarquista"):
    
    # Iterar sobre los dos periodos
    for period_name, (start_year, end_year) in PERIODOS.items():
        
        # 1. Filtrar el DataFrame para el período
        # Incluimos 2007 en ambos para ver la línea de tendencia completa.
        df_period = df_group[(df_group["Año"] >= start_year) & (df_group["Año"] <= end_year)].copy()
        
        # Saltar si el período no tiene suficientes datos
        if df_period.empty or len(df_period) < 2:
            continue
            
        x = df_period["Año"]
        y = df_period["Gasto_per_pea"]
        
        # 2. Ajuste lineal
        coef = np.polyfit(x, y, 1)
        pol = np.poly1d(coef)

        # 3. Recta: Solo dos puntos para dibujar la línea de regresión de inicio a fin del período
        x_line = np.array([start_year, end_year]) 
        y_line = pol(x_line)

        # 4. Variación promedio (final de la recta vs inicio de la recta)
        # y_line[1] es el valor predicho al final del período, y_line[0] al inicio.
        var_pct = ((y_line[1] / y_line[0] - 1) * 100)
        
        # 5. Traza de la recta
        fig.add_trace(go.Scatter(
            x=x_line,
            y=y_line,
            mode="lines",
            name=f"Reg. {grupo} {period_name}",
            line=dict(dash="dash", width=2)
        ))

        # 6. Anotación
        # Posicionar el texto en el centro de su período (ej. 2001 para P1, 2016 para P2)
        ann_x = 2001 if start_year == 1995 else 2016
        ann_y = pol(ann_x)
        
        # Ajustar la posición Y: Si la pendiente (coef[0]) es positiva, colocar el texto un 2% arriba; si es negativa, un 2% abajo.
        y_placement = ann_y * (1.02 if coef[0] > 0 else 0.98) 

        fig.add_annotation(
            x=ann_x,
            y=y_placement, 
            text=f"Var {period_name}: {var_pct:+.0f}%",
            showarrow=False,
            font=dict(size=14, color="black")
        )

# Final layout updates
fig.update_layout(
    height =800,
    title="Evolución anual del Gasto per PEA agrupado por Clasificacion Minarquista por Periodo",
    xaxis=dict(dtick=1,
               range=[1994.5, 2023.5]),
    yaxis=dict(
            title="Gasto per PEA (€/hab)",
            range = [1000, 20000]),
    yaxis2=dict(title="PIB per PEA",
                overlaying="y",
                side="right",
                range = [10000, 70000]),
    legend_title_text='Clasificación / Regresión'
)

fig.show()
```
